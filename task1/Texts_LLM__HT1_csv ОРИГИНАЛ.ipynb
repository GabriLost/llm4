```python
import gensim.downloader as api
from sklearn.metrics.pairwise import cosine_similarity

model = api.load("word2vec-google-news-300") # или другую, подходящую для русского языка
```

```python
def get_average_vector(text, model):
    words = text.split()  # разделяем текст на слова
    word_vectors = []
    for word in words:
        if word in model:  # проверяем, есть ли слово в модели
            word_vectors.append(model[word])
    if len(word_vectors) == 0:
        return np.zeros(model.vector_size)  # если нет слов в модели, возвращаем нулевой вектор
    return np.mean(word_vectors, axis=0)
```
